```python
from llama_index.core import VectorStoreIndex , SimpleDirectoryReader, Settings

from llama_index.llms.groq import Groq
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
```

```python
class RAGPipeline:
    def __init__(self):
        llm_model= ""
        api_key = ""
        data_dir = "/home/vishal/repos/data/Sent_files"
        embed_model = "local:BAAI/bge-small-en-v1.5"
        
        self.embed_model = embed_model
        self.llm = Groq(model = llm_model,
                       api_key = api_key
                       )
        self.data_dir = data_dir
        self.index = None
        self.query_engine = None
        self._setup()

    def _setup(self):
        Settings.llm = self.llm
        Settings.embed_model = self.embed_model

        ## Chunking sementic chunking
        from llama_index.core.node_parser import SemanticSplitterNodeParser
        embed_model = HuggingFaceEmbedding(model_name = "BAAI/bge-small-en-v1.5")
        Settings.text_splitter = SemanticSplitterNodeParser(embed_model = embed_model)

        documents = SimpleDirectoryReader(self.data_dir).load_data()
        self.index = VectorStoreIndex.from_documents(documents)

        self.query_engine = self.index.as_query_engine(
            similarity_top_k = 3 ,
            responce_mode = "compect"
        )
    def responce(self, question):
        return self.query_engine.query(question)
```

```python
rag = RAGPipeline()
```

```python
responce = rag.responce("Explain which system you are using , whic model,")
print(responce)
```
