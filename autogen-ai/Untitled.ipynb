```python
planner = AssistantAgent(
    name = "planner",
    system_message = "Think, plan and decide actions",
    code_execution_config = {"use_docker" : False}
)

executor = UserProxyAgent(
    name = "executer",
    human_input_mode = "NEVER",
    max_consecutive_auto_reply = 5,
    is_termination_msg = lambda x :"Done" in x["content"],
    code_execution_config={"use_docker": False} 
)
```

```python
executor.initiate_chat(
    planner,
    message = "write python code to add two numbers, execute it , and say Done" 
)
```

```python
from autogen import UserProxyAgent

executor = UserProxyAgent(
    name="executor",
    human_input_mode="NEVER",
    code_execution_config={
        "use_docker": False
    }
)

task_code = 'print("Hello, AutoGen")'

exit_code, stdout, stderr = executor.run_code(task_code, use_docker=False)

print("Exit code:", exit_code)
print("STDOUT:", stdout)
print("STDERR:", stderr)
```
